## 1. Creating Memory Instances

Let's start by creating instances of each memory type:

```python
import os
from agentmem.semantic import SemanticMemory
from agentmem.episodic import EpisodicMemory
from agentmem.procedural import ProceduralMemory

# Create a directory for persistence
os.makedirs("./notebook_memory", exist_ok=True)

# Create memory instances
semantic_memory = SemanticMemory(
    persistence="./notebook_memory",
    vector_search=True
)

episodic_memory = EpisodicMemory(
    persistence="./notebook_memory",
    vector_search=True
)

procedural_memory = ProceduralMemory(
    persistence="./notebook_memory",
    vector_search=True
)

print("Memory instances created successfully.")
```

## 2. Exploring Semantic Memory

Semantic memory stores factual knowledge and concepts. Let's create some facts and explore how to query them.

```python
# Add some geographic facts
paris_id = semantic_memory.create(
    content="Paris is the capital city of France and is known for the Eiffel Tower.",
    category="geography",
    tags=["europe", "cities", "capitals", "france"]
)

tokyo_id = semantic_memory.create(
    content="Tokyo is the capital city of Japan and is the world's most populous metropolitan area.",
    category="geography",
    tags=["asia", "cities", "capitals", "japan"]
)

# Add some programming-related facts
python_id = semantic_memory.create(
    content="Python is a high-level programming language known for its readability and versatility.",
    category="programming",
    tags=["languages", "python", "software"]
)

javascript_id = semantic_memory.create(
    content="JavaScript is a scripting language primarily used for web development.",
    category="programming",
    tags=["languages", "javascript", "web", "software"]
)

print(f"Created facts with IDs: {paris_id}, {tokyo_id}, {python_id}, {javascript_id}")
```

### Reading Facts

Now, let's read back one of the facts we just created:

```python
# Read the Paris fact
paris_fact = semantic_memory.read(paris_id)

# Display it nicely
print(f"Content: {paris_fact['content']}")
print(f"Category: {paris_fact['category']}")
print(f"Tags: {', '.join(paris_fact['tags'])}")
print(f"Created at: {paris_fact['created_at']}")
```

### Updating Facts

Let's update one of our facts with additional information:

```python
# Update the Paris fact
semantic_memory.update(
    paris_id,
    content="Paris is the capital city of France and is known for the Eiffel Tower, the Louvre Museum, and Notre-Dame Cathedral.",
    tags=["europe", "cities", "capitals", "france", "tourism"]
)

# Read the updated fact
updated_paris = semantic_memory.read(paris_id)
print(f"Updated content: {updated_paris['content']}")
print(f"Updated tags: {', '.join(updated_paris['tags'])}")
```

### Querying Semantic Memory

Now let's try various ways to query our semantic memory:

```python
# Simple keyword search
capital_results = semantic_memory.query("capital")
print(f"Found {len(capital_results)} results for 'capital':")
for i, result in enumerate(capital_results, 1):
    print(f"{i}. {result['content'][:60]}...")

print("\n")

# Search by category
programming_results = semantic_memory.query("", category="programming")
print(f"Found {len(programming_results)} results in category 'programming':")
for i, result in enumerate(programming_results, 1):
    print(f"{i}. {result['content'][:60]}...")

print("\n")

# Search by tags
language_results = semantic_memory.query("", tags=["languages"])
print(f"Found {len(language_results)} results with tag 'languages':")
for i, result in enumerate(language_results, 1):
    print(f"{i}. {result['content'][:60]}...")
```

### Vector Search

Let's try a vector search to find semantically similar content, even when exact keywords don't match:

```python
# Vector search for concepts related to "tourist attractions"
tourist_results = semantic_memory.query(
    "famous tourist attractions in Europe",
    use_vector=True,
    n_results=2
)

print(f"Vector search results for 'famous tourist attractions in Europe':")
for result in tourist_results:
    print(f"Content: {result['content']}")
    print(f"Similarity score: {result.get('similarity_score', 'N/A')}")
    print("---")

# Try another vector search for programming concepts
coding_results = semantic_memory.query(
    "software development technologies",
    use_vector=True,
    n_results=2
)

print(f"\nVector search results for 'software development technologies':")
for result in coding_results:
    print(f"Content: {result['content']}")
    print(f"Similarity score: {result.get('similarity_score', 'N/A')}")
    print("---")
```

## 3. Exploring Episodic Memory

Episodic memory stores experiences and events with temporal context. Let's create some episodic memories.

```python
from datetime import datetime, timedelta

# Current time
now = datetime.now()

# Create memories at different times
episode1_id = episodic_memory.create(
    content="User asked about how to sort a list in Python",
    timestamp=now - timedelta(days=1),  # Yesterday
    context={
        "user_id": "user123",
        "conversation_id": "conv456",
        "topic": "python-basics"
    },
    importance=6
)

episode2_id = episodic_memory.create(
    content="Explained lambda functions and list comprehensions in Python",
    timestamp=now - timedelta(hours=12),  # 12 hours ago
    context={
        "user_id": "user123",
        "conversation_id": "conv457",
        "topic": "python-intermediate"
    },
    importance=7
)

episode3_id = episodic_memory.create(
    content="User expressed confusion about JavaScript promises and async/await",
    timestamp=now - timedelta(hours=6),  # 6 hours ago
    context={
        "user_id": "user456",
        "conversation_id": "conv458",
        "topic": "javascript-advanced",
        "sentiment": "confused"
    },
    importance=8
)

print(f"Created episodic memories with IDs: {episode1_id}, {episode2_id}, {episode3_id}")
```

### Reading Episodic Memories

```python
# Read an episodic memory
episode = episodic_memory.read(episode2_id)

# Display it nicely
print(f"Content: {episode['content']}")
print(f"Timestamp: {episode['timestamp']}")
print(f"Context:")
for key, value in episode['context'].items():
    print(f"  - {key}: {value}")
print(f"Importance: {episode['importance']}")
```

### Querying Episodic Memory by Time and Context

```python
# Query by time range
last_12_hours = episodic_memory.query(
    "",  # Empty query matches all content
    start_time=now - timedelta(hours=12),
    end_time=now
)

print(f"Experiences from the last 12 hours:")
for exp in last_12_hours:
    print(f"- {exp['content']}")
    print(f"  Time: {exp['timestamp']}")

print("\n")

# Query by context keys and minimum importance
important_user_exp = episodic_memory.query(
    "",
    context_keys=["user_id", "sentiment"],  # Must have these keys
    min_importance=7  # At least importance level 7
)

print(f"Important experiences with user_id and sentiment:")
for exp in important_user_exp:
    print(f"- {exp['content']}")
    print(f"  Importance: {exp['importance']}")
    print(f"  Context: {exp['context']}")
```

### Content Search and Vector Search in Episodic Memory

```python
# Content keyword search
python_experiences = episodic_memory.query("Python")
print(f"Experiences mentioning Python:")
for exp in python_experiences:
    print(f"- {exp['content']}")

print("\n")

# Vector search for semantically similar experiences
vector_results = episodic_memory.query(
    "User needed help with programming functions",
    use_vector=True,
    n_results=2
)

print(f"Vector search results:")
for result in vector_results:
    print(f"Content: {result['content']}")
    print(f"Similarity score: {result.get('similarity_score', 'N/A')}")
    print("---")
```

## 4. Exploring Procedural Memory

Procedural memory stores knowledge about how to perform tasks. Let's create some procedural memories.

```python
# Create procedural memories for different tasks
sort_list_id = procedural_memory.create(
    content="How to sort a list in Python",
    task="Sort a Python list",
    steps=[
        "Use the built-in sort() method: my_list.sort()",
        "For custom sorting, use sort() with a key function: my_list.sort(key=func)",
        "For descending order, use reverse=True: my_list.sort(reverse=True)",
        "To get a new sorted list instead of modifying the original, use sorted(): new_list = sorted(my_list)"
    ],
    prerequisites=["Basic Python knowledge", "Understanding of lists"],
    domains=["programming", "python", "data-structures"]
)

api_request_id = procedural_memory.create(
    content="How to make an API request in JavaScript",
    task="Make an API request",
    steps=[
        "Using fetch API: fetch('https://api.example.com/data')",
        "Handle the response with .then(): .then(response => response.json())",
        "Process the data: .then(data => console.log(data))",
        "Catch errors: .catch(error => console.error('Error:', error))",
        "Alternative: Use async/await with try/catch blocks for cleaner code"
    ],
    prerequisites=["JavaScript knowledge", "Understanding of Promises"],
    domains=["programming", "javascript", "web-development"]
)

git_commit_id = procedural_memory.create(
    content="How to create a Git commit",
    task="Create a Git commit",
    steps=[
        "Stage changes: git add <filename> or git add . for all changes",
        "Create commit with message: git commit -m "Your commit message"",
        "To stage and commit in one step: git commit -am "Your commit message"",
        "Push to remote repository: git push origin <branch-name>"
    ],
    prerequisites=["Git installed", "Repository initialized"],
    domains=["version-control", "git", "development-workflow"]
)

print(f"Created procedural memories with IDs: {sort_list_id}, {api_request_id}, {git_commit_id}")
```

### Reading Procedural Memories

```python
# Read a procedural memory
procedure = procedural_memory.read(git_commit_id)

# Display it nicely
print(f"Task: {procedure['task']}")
print(f"Content: {procedure['content']}")
print(f"Steps:")
for i, step in enumerate(procedure['steps'], 1):
    print(f"  {i}. {step}")
print(f"Prerequisites: {', '.join(procedure['prerequisites'])}")
print(f"Domains: {', '.join(procedure['domains'])}")
```

### Querying Procedural Memory

```python
# Query by domain
python_procedures = procedural_memory.query(
    "",
    domain="python"
)

print(f"Python-related procedures:")
for proc in python_procedures:
    print(f"- {proc['task']}")

print("\n")

# Query by content keywords
api_procedures = procedural_memory.query("API")
print(f"Procedures mentioning API:")
for proc in api_procedures:
    print(f"- {proc['task']}")
    print(f"  Steps: {len(proc['steps'])} steps")

print("\n")

# Vector search for procedures
version_control_procedures = procedural_memory.query(
    "How to save my code changes",
    use_vector=True,
    n_results=2
)

print(f"Vector search results for 'How to save my code changes':")
for proc in version_control_procedures:
    print(f"Task: {proc['task']}")
    print(f"Similarity score: {proc.get('similarity_score', 'N/A')}")
    print("---")
```

## 6. Integrating Memory Types

Now, let's build a simple agent that uses all three memory types to showcase how they can work together.

```python
class MemoryAgent:
    def __init__(self, name="Memory Agent", persistence_dir=None):
        self.name = name
        
        # Initialize memory systems
        self.semantic = SemanticMemory(
            persistence=persistence_dir,
            vector_search=True
        )
        
        self.episodic = EpisodicMemory(
            persistence=persistence_dir,
            vector_search=True
        )
        
        self.procedural = ProceduralMemory(
            persistence=persistence_dir,
            vector_search=True
        )
        
        # Load memories if persistence is enabled
        if persistence_dir:
            self.semantic.load_all()
            self.episodic.load_all()
            self.procedural.load_all()
    
    def memorize_fact(self, fact, category=None, tags=None):
        """Store a fact in semantic memory"""
        fact_id = self.semantic.create(
            content=fact,
            category=category or "general",
            tags=tags or []
        )
        print(f"Memorized fact: {fact}")
        return fact_id
    
    def record_experience(self, experience, context=None, importance=5):
        """Record an experience in episodic memory"""
        experience_id = self.episodic.create(
            content=experience,
            timestamp=datetime.now(),
            context=context or {},
            importance=importance
        )
        print(f"Recorded experience: {experience}")
        return experience_id
    
    def learn_procedure(self, procedure, task, steps, domains=None):
        """Store a procedure in procedural memory"""
        procedure_id = self.procedural.create(
            content=procedure,
            task=task,
            steps=steps,
            domains=domains or ["general"]
        )
        print(f"Learned procedure: {task}")
        return procedure_id
    
    def ask(self, query):
        """Answer a question using all memory types"""
        print(f"\nProcessing query: {query}")
        
        # Record this interaction
        self.record_experience(
            f"Received query: {query}",
            context={"type": "question", "timestamp": datetime.now().isoformat()}
        )
        
        # Check semantic memory for facts
        facts = self.semantic.query(query, use_vector=True, n_results=2)
        
        # Check procedural memory for how-to information
        procedures = self.procedural.query(query, use_vector=True, n_results=2)
        
        # Check episodic memory for similar past interactions
        past_experiences = self.episodic.query(query, use_vector=True, n_results=2)
        
        # Generate a response using the retrieved memories
        response = f"Here's what I know about '{query}':\n\n"
        
        if facts:
            response += "Facts:\n"
            for fact in facts:
                response += f"- {fact['content']}\n"
            response += "\n"
        
        if procedures:
            response += "Procedures:\n"
            for proc in procedures:
                response += f"- {proc['task']}:\n"
                for i, step in enumerate(proc['steps'], 1):
                    response += f"  {i}. {step}\n"
            response += "\n"
        
        if past_experiences:
            filtered_experiences = [exp for exp in past_experiences 
                                  if exp['context'].get('type') != 'question']
            if filtered_experiences:
                response += "Related past experiences:\n"
                for exp in filtered_experiences:
                    response += f"- {exp['content']}\n"
        
        if not facts and not procedures and not filtered_experiences:
            response += "I don't have specific information about that yet."
        
        # Record this response
        self.record_experience(
            f"Provided response about: {query}",
            context={"type": "response", "timestamp": datetime.now().isoformat()}
        )
        
        return response
    
    def save(self):
        """Save all memories"""
        self.semantic.save_all()
        self.episodic.save_all()
        self.procedural.save_all()
```

Let's create an agent and populate it with memories:

```python
# Create a memory agent
agent = MemoryAgent(name="Tutorial Agent", persistence_dir="./agent_memory")

# Add some facts
agent.memorize_fact(
    "Python was created by Guido van Rossum in 1991.",
    category="programming",
    tags=["python", "history"]
)

agent.memorize_fact(
    "The Eiffel Tower is 330 meters tall and was completed in 1889.",
    category="landmarks",
    tags=["paris", "france", "architecture"]
)

# Add some procedures
agent.learn_procedure(
    "How to install Python packages",
    "Install a Python package",
    [
        "Use pip install: pip install package-name",
        "For a specific version: pip install package-name==1.2.3",
        "From requirements file: pip install -r requirements.txt",
        "In development mode: pip install -e ."
    ],
    domains=["python", "package-management"]
)

# Record some experiences
agent.record_experience(
    "User expressed interest in visiting the Eiffel Tower",
    context={"topic": "travel", "location": "paris", "type": "interest"}
)

agent.record_experience(
    "Explained how to use Python's list comprehension feature",
    context={"topic": "python", "feature": "list comprehension", "type": "explanation"}
)
```

Now let's interact with our agent:

```python
# Ask questions to the agent
questions = [
    "Tell me about Python",
    "How do I install packages?",
    "What can you tell me about Paris?"
]

for question in questions:
    answer = agent.ask(question)
    print(f"Question: {question}")
    print(f"Answer:\n{answer}")
    print("\n" + "-"*50 + "\n")
```

Let's save the agent's state:

```python
# Save all memories
agent.save()
print("Agent state saved to disk.")
```

## 7. Advanced Features

Let's explore some advanced features of AgentMem, such as concurrency management and memory metrics.

```python
from agentmem.concurrency import lock_manager
from agentmem.logging import get_metrics_collector, get_memory_tracker

# Enable lock monitoring
lock_manager.enable_monitoring()

# Get metrics collector and memory tracker
metrics = get_metrics_collector()
memory_tracker = get_memory_tracker()
```

### Concurrency Example

Let's demonstrate how AgentMem handles concurrent operations:

```python
import threading
import time

# Function that will run in multiple threads
def worker(name, memory, iterations):
    print(f"Worker {name} starting")
    for i in range(iterations):
        # Create a memory
        memory_id = memory.create(
            content=f"Memory from worker {name}, iteration {i}",
            category="test"
        )
        
        # Read it back
        entry = memory.read(memory_id)
        
        # Update it
        memory.update(
            memory_id,
            content=f"Updated memory from worker {name}, iteration {i}"
        )
        
        # Small delay to simulate work
        time.sleep(0.01)
    
    print(f"Worker {name} completed")

# Create a test memory instance
test_memory = SemanticMemory()

# Create multiple threads
threads = []
for i in range(3):
    t = threading.Thread(target=worker, args=(f"Thread-{i}", test_memory, 5))
    threads.append(t)
    t.start()

# Wait for all threads to complete
for t in threads:
    t.join()

print("All workers completed")

# Check lock statistics
stats = lock_manager.get_lock_statistics()
print("\nLock Statistics:")
print(f"Total lock acquisitions: {stats['total_acquisitions']}")
print(f"Total lock releases: {stats['total_releases']}")
print(f"Lock contentions: {stats['contentions']}")
print(f"Average lock hold time: {stats['avg_hold_time']:.6f} seconds")

# Get the results
results = test_memory.query("", category="test")
print(f"\nCreated {len(results)} memories in total")
```

### Memory Usage Metrics

```python
# Get memory usage stats
usage = memory_tracker.get_memory_usage()
print("Memory Usage:")
for memory_type, size in usage.items():
    print(f"{memory_type}: {size / 1024:.2f} KB")

# Get operation metrics
operation_stats = metrics.get_operation_stats()
print("\nOperation Counts:")
for op_name, stats in operation_stats.items():
    print(f"{op_name}: {stats.get('count', 0)} operations, avg time: {stats.get('avg_time', 0):.6f} seconds")
```
